## 1. A notebook is a stack of cells

A notebook is made of **cells**, stacked from top to bottom. There are two kinds:

- **Markdown cells** hold formatted text — like the one you are reading now.
- **Code cells** hold Python code that the computer can run.

You can tell them apart at a glance: a code cell has a `[ ]` on its left and a slightly
different background. The cell just below this one is a **code cell**.


````python
# This is a code cell.
# Lines starting with a # are comments — notes for humans, ignored by Python.
# The line below is real code. Run this cell to see what it does.
print("Hello from your first code cell!")
````

## 5. The order you run cells in matters

This is the one idea about notebooks that surprises everybody, so it is worth meeting it
deliberately. A notebook remembers things from every cell you have run, in the order you ran
them — **not** in the order the cells appear on the screen.

Run the next three cells **in order**, top to bottom.


````python
message = "first value"
````

````python
print(message)
````

````python
message = "second value"
````

Now scroll back up and run the **middle** cell (`print(message)`) again, without
running the others. It now prints `second value`, even though the cell above it still says
`first value` on screen. The notebook remembers the *last* value you gave `message`, which
came from the cell further down.

This is why, when something behaves strangely, the fix is almost always the same:

> **Restart the kernel and run every cell from the top.**

## 6. The kernel: the notebook's memory

Behind your notebook runs a **kernel** — the Python program that actually executes your
cells and remembers your variables. When people say *"restart the kernel"*, they mean:
throw away everything the notebook currently remembers and start fresh.

You will do this often, and you should do it now to see that it is harmless.

**Do this:**

1. In the menu, find **Kernel &gt; Restart Kernel** (in VS Code: the *Restart* button at the
   top of the notebook).
2. After it restarts, run the cell below **on its own**, without running anything above it
   first.


````python
print(city)
````

You will very likely get a red **`NameError`** saying `name 'city' is not defined`.
That is expected and correct: restarting the kernel wiped the notebook's memory, so `city`
no longer exists until you run the cell that creates it again.

**Now fix it the standard way:** use the menu **Run &gt; Run All Cells** (or *Run All* at the
top). Every cell runs from the top, `city` is created again, and the error disappears.

Remember this cure — you will use it all semester:

> Something is behaving oddly? **Restart the kernel, then Run All.**

## What you learned today

- A notebook is a stack of **markdown** cells (text) and **code** cells (Python).
- **`Shift + Enter`** runs a cell; output appears underneath, and the `[n]` counts runs.
- The notebook remembers values in the **order you run cells**, not top-to-bottom on screen.
- When things get confusing: **restart the kernel, then Run All**.
- `=` stores a value in a variable; **strings** use quotation marks, **numbers** do not.
- `print(...)` shows text and variables; commas put spaces between the parts.
