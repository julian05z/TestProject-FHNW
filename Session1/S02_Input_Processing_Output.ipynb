## 2.2 From the problem to a testable design

Before writing code, describe who needs what, how success can be observed, and which data must flow through the program.


A **User Story** consists of the **role**, **capability**, and **benefit**:
> As a **[role]**,  
> I want **[capability]**,  
> so that **[benefit]**.


**Acceptance criteria** can be expressed as concrete examples using the **Given-When-Then** structure:
> **Given** describes the initial situation and relevant inputs.  
> **When** describes the action or event being performed.  
> **Then** describes the expected result.  


Example:
> As a **customer**,  
> I want to **see the total price of my order**,  
> so that I **know how much I need to pay**.

> **Given** a pizza with a unit price of CHF 12.50 and a quantity of 2,  
> **when** the order total is calculated,  
> **then** the resulting total is CHF 25.00.

## 2.3 Input - Processing - Output

All Information Systems, Programs, AI Systems, etc follow three fundamental steps:

1. Input
    * Any data that the program receives while it is running
2. Processing
    * Any action the program does while it is running to create value.
3. Output
    * Any data the program produced while it is running.

<br>  
  
| Component     | Guiding question                    | Pizza example                   |
| ------------- | ----------------------------------- | ------------------------------- |
| 1. Input      | What data does the program receive? | Unit price and quantity         |
| 2. Processing | What operation is performed?        | Multiply unit price by quantity |
| 3. Output     | What result is produced?            | Order total                     |
