## 3.1 Functions

A function definition names a reusable block of instructions. Parameters receive the argument values supplied in a call. `return` sends a value back to the caller; `print()` displays text.

```python
def delivery_fee(distance):
    return 2.0 + distance * 0.5

fee = delivery_fee(4)
print(f"CHF {fee:.2f}")  # CHF 4.00
```

A function can have no parameters. Without an explicit return value, Python returns `None`.

## 3.4 Data structures

| Structure | Main properties | Example |
|---|---|---|
| `list` | Ordered, mutable, permits duplicates | A changing basket |
| `tuple` | Ordered, immutable, permits duplicates | Coordinates |
| `set` | Unordered, mutable, unique values | Product tags |
| `dict` | Key–value pairs, mutable, unique keys, preserves insertion order | Prices by product name |

**Ordered** means items have defined positions; it does not mean sorted by value. A tuple's items cannot be added, removed, or replaced. A dictionary can contain the same value under different keys.

Small syntax examples:
```python
colours = ["red", "blue"]
colours.append("green")
first_colour = colours[0]      # indexes start at zero
point = (4, 8)
x = point[0]
tags = {"new", "local"}
tags.add("new")                # duplicates do not add another entry
is_local = "local" in tags
prices = {"tea": 3.0}
prices["tea"] = 3.5
tea_price = prices["tea"]
```
